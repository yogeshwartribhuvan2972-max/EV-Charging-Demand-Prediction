## Notebook 05 : Charging Stations Data Preprocessing

### Business Objective

The objective of this notebook is to clean and preprocess the charging stations dataset. This dataset will be used to analyze charging station coverage, identify underserved locations, and support charging station expansion recommendations.

```python
# ==========================================================
# Import Required Libraries
# ==========================================================

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

```python
# ==========================================================
# Load Charging Stations Dataset
# ==========================================================

stations = pd.read_excel("../data/raw/charging_stations.xlsx")

print("Dataset Loaded Successfully")
```

```python
print("Shape :", stations.shape)

stations.head()
```

```python
stations.columns.tolist()
```

```python
stations.info()
```

```python
stations.describe(include="all")
```

```python
print("Missing Values")
print(stations.isnull().sum())
```

```python
print("Duplicate Records :", stations.duplicated().sum())
```

```python
print(stations.dtypes)
```

```python
text_columns = stations.select_dtypes(include="object").columns

for col in text_columns:
    stations[col] = stations[col].str.strip()
```

```python
for col in text_columns:
    stations[col] = stations[col].str.title()
```

```python
print("Unique Years :", stations["Year"].nunique())

print("Unique States :", stations["State"].nunique())

print("Unique Cities :", stations["City"].nunique())
```

```python
stations.to_csv(
    "../data/processed/charging_stations_clean.csv",
    index=False
)

print("Charging Stations Dataset Saved Successfully")
```

## Notebook 04 : Station Usage Data Preprocessing

### Business Objective

The objective of this notebook is to clean and preprocess the station usage dataset. This dataset will be used to analyze charging sessions, waiting time, station utilization, and charging demand for business and customer recommendations.

```python
# ==========================================================
# Import Required Libraries
# ==========================================================

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

```python
# ==========================================================
# Load Dataset
# ==========================================================

usage = pd.read_excel("../data/raw/station_usage.xlsx")

print("Dataset Loaded Successfully")
```

```python
print("Shape :", usage.shape)

usage.head()
```

```python
usage.columns.tolist()
```

```python
usage.info()
```

```python
usage.describe(include="all")
```

```python
print("Missing Values")

print(usage.isnull().sum())
```

```python
print("Duplicate Records :", usage.duplicated().sum())
```

```python
print(usage.dtypes)
```

```python
text_columns = usage.select_dtypes(include="object").columns

for col in text_columns:
    usage[col] = usage[col].str.strip()
```

```python
for col in text_columns:
    usage[col] = usage[col].str.title()
```

```python
print("Unique Stations :", usage["Station_ID"].nunique())

print("\nDemand Level")
print(usage["Demand_Level"].value_counts())

print("\nYears")
print(usage["Year"].unique())

print("\nMonths")
print(sorted(usage["Month"].unique()))
```

```python
usage.to_csv(
    "../data/processed/station_usage_clean.csv",
    index=False
)

print("Station Usage Dataset Saved Successfully")
```

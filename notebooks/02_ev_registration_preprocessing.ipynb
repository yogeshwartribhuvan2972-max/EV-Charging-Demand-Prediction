## Notebook 02 : EV Registration Data Preprocessing

### Business Objective

The objective of this notebook is to clean and preprocess the EV registration dataset. This dataset will be used to analyze EV adoption trends, estimate future EV growth, and support charging demand prediction.

```python
# ==========================================================
#   Import Required Libraries
# ==========================================================

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

pd.set_option("display.max_columns", None)
```

```python
# ==========================================================
#   Load EV Registration Dataset
# ==========================================================

ev = pd.read_excel("../data/raw/ev_registration.xlsx")

print("Dataset Loaded Successfully")
```

```python
ev.rename(columns={
    "District/City":"City"
}, inplace=True)
```

```python
ev["State"] = ev["State"].str.strip()
ev["City"] = ev["City"].str.strip()
```

```python
ev["State"] = ev["State"].str.title()
ev["City"] = ev["City"].str.title()
ev["Month"] = ev["Month"].str.title()
```

```python
numeric_cols = [
    "Year",
    "2W",
    "3W",
    "4W",
    "Bus",
    "Goods"
]

ev[numeric_cols] = ev[numeric_cols].apply(pd.to_numeric)
```

```python
ev["Total_EV"] = (
    ev["2W"] +
    ev["3W"] +
    ev["4W"] +
    ev["Bus"] +
    ev["Goods"]
)
```

```python
month_map = {
    "January":1,
    "February":2,
    "March":3,
    "April":4,
    "May":5,
    "June":6,
    "July":7,
    "August":8,
    "September":9,
    "October":10,
    "November":11,
    "December":12
}

ev["Month_No"] = ev["Month"].map(month_map)
```

```python
ev = ev[
[
"Year",
"Month",
"Month_No",
"State",
"City",
"2W",
"3W",
"4W",
"Bus",
"Goods",
"Total_EV"
]
]
```

## Notebook 03 : Station Master Data Preprocessing

### Business Objective

The objective of this notebook is to clean and preprocess the charging station master dataset. This dataset contains charging station information that will be used for charging demand analysis, waiting time prediction, station utilization analysis, and business recommendations.

```python
# ==========================================================
# Import Required Libraries
# ==========================================================

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

```python
# ==========================================================
# Load Station Master Dataset
# ==========================================================

station = pd.read_excel("../data/raw/station_master.xlsx")

print("Dataset Loaded Successfully")
```

```python
text_columns = station.select_dtypes(include="object").columns

for col in text_columns:
    station[col] = station[col].str.strip()
```

```python
for col in text_columns:
    station[col] = station[col].str.title()
```

```python
# ==========================================================
# Check Unique Values
# ==========================================================

print("Unique States :", station["State"].nunique())
print("Unique Cities :", station["City"].nunique())

print("\nOperator Distribution")
print(station["Operator"].value_counts())

print("\nStatus Distribution")
print(station["Status"].value_counts())

print("\nStation Type Distribution")
print(station["Station_Type"].value_counts())

print("\nCharger Type Distribution")
print(station["Charger_Type"].value_counts())
```

```python
station["Fast_Charger_Ratio"] = (
    station["Fast_Chargers"] /
    station["Total_Chargers"]
)
```

```python
station["DC_Ratio"] = (
    station["DC_Chargers"] /
    station["Total_Chargers"]
)
```

```python
station["Station_Age"] = (
    2025 -
    station["Installation_Year"]
)
```

```python
station["Utilization_Level"] = pd.cut(
    station["Nearby_EV_Population"],
    bins=[0,5000,15000,30000,100000],
    labels=["Low","Medium","High","Very High"]
)
```

```python
station.to_csv(
    "../data/processed/station_master_clean.csv",
    index=False
)
```

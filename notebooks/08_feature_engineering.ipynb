# EV Charging Infrastructure Planning & Demand Prediction

## Notebook 08 : Feature Engineering

### Business Objective

The objective of this notebook is to prepare the master dataset for machine learning by handling missing values, encoding categorical variables, creating new features, selecting important features, and saving the final model-ready dataset.

```python
# ==========================================================
# Import Required Libraries
# ==========================================================

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder

pd.set_option("display.max_columns", None)
```

```python
# ==========================================================
# Load Master Dataset
# ==========================================================

master = pd.read_csv("../data/processed/master_dataset.csv")

print("Dataset Loaded Successfully")
```

```python
print(master.shape)

master.head()
```

```python
print(master.isnull().sum())
```

```python
numeric_columns = master.select_dtypes(include=np.number).columns

master[numeric_columns] = master[numeric_columns].fillna(0)
```

```python
categorical_columns = master.select_dtypes(include="object").columns

master[categorical_columns] = master[categorical_columns].fillna("Unknown")
```

```python
print(master.isnull().sum())
```

```python
master["EV_Per_Station"] = (
    master["Total_EV"] /
    (master["Total_Charging_Stations"] + 1)
)
```

```python
master["Chargers_Per_Station"] = (
    master["Total_Chargers"] /
    (master["Total_Charging_Stations"] + 1)
)
```

```python
master["Revenue_Per_Session"] = (
    master["Revenue_INR"] /
    (master["Sessions"] + 1)
)
```

```python
master["Completion_Rate"] = (
    master["Completed_Sessions"] /
    (master["Sessions"] + 1)
)
```

```python
master["Cancellation_Rate"] = (
    master["Cancelled_Sessions"] /
    (master["Sessions"] + 1)
)
```

```python
master["Downtime_Percentage"] = (
    master["Downtime_Hours"] /
    720
) * 100
```

```python
master["Queue_Per_User"] = (
    master["Queue_Length"] /
    (master["Avg_Daily_Users"] + 1)
)
```

```python
categorical_columns = master.select_dtypes(include="object").columns

encoder = LabelEncoder()

for col in categorical_columns:
    master[col] = encoder.fit_transform(master[col])
```

```python
master.to_csv(
    "../data/processed/model_dataset.csv",
    index=False
)

print("="*50)
print("Feature Engineering Completed Successfully")
print("="*50)
print("Final Shape :", master.shape)
```

```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

pd.set_option("display.max_columns",None)
```

```python
stations_df = pd.read_csv("../data/raw/ev_charging_stations.csv")

usage_df = pd.read_csv("../data/raw/ev_charging_usage.csv")

registration_df = pd.read_csv("../data/raw/ev_registration_by_state.csv")

population_df = pd.read_csv("../data/raw/indian_cities_population.csv")
```

```python
datasets = {
    "EV Charging Stations": stations_df,
    "EV Charging Usage": usage_df,
    "EV Registration": registration_df,
    "Population": population_df
}

overview = pd.DataFrame({
    "Dataset":[name for name in datasets],
    "Rows":[df.shape[0] for df in datasets.values()],
    "Columns":[df.shape[1] for df in datasets.values()]
})

overview
```

```python
for name,df in datasets.items():

    print("="*70)
    print(name)
    print("="*70)

    display(df.dtypes)
```

```python
for name,df in datasets.items():

    print("="*70)
    print(name)
    print("="*70)

    print("Duplicate Records :",df.duplicated().sum())
```

```python
for name,df in datasets.items():

    print("="*70)
    print(name)
    print("="*70)

    display(df.isnull().sum())
```

```python
for name,df in datasets.items():

    print("="*70)
    print(name)
    print("="*70)

    display(df.nunique())
```

# EV Charging Stations Dataset

## Dataset Description

The EV Charging Stations dataset contains information about existing electric vehicle charging stations across different cities and states in India. It provides geographical details, charging operators, connector types, charging capacity, and usage categories.

## Business Importance

This dataset helps understand the current charging infrastructure available across India. It is useful for analyzing charging station distribution, operator presence, connector availability, and identifying regions where additional charging stations may be required.

## Expected Use in the Project

- Charging Station Analysis
- Operator Analysis
- State-wise Infrastructure Analysis
- Interactive Map Visualization
- Feature Engineering

# EV Charging Usage Dataset

## Dataset Description

This dataset contains historical charging session information recorded at EV charging stations. It includes charging duration, energy consumed, charging cost, payment method, and vehicle type.

## Business Importance

The dataset helps understand charging behavior, energy consumption patterns, charging demand, and customer usage trends.

## Expected Use in the Project

- Demand Analysis
- Revenue Analysis
- Peak Usage Analysis
- Energy Consumption Analysis
- Demand Prediction

# EV Registration by State Dataset

## Dataset Description

This dataset contains state-wise electric vehicle registration information across India. It represents the adoption of electric vehicles in different states.

## Business Importance

The dataset helps identify states with higher EV adoption and supports market analysis and business expansion planning.

## Expected Use in the Project

- State-wise EV Analysis
- Market Analysis
- Business Dashboard
- Expansion Planning

**Note:** This dataset is used for business analytics and visualization. Since the data is available at the state level, it is not directly used for city-level demand prediction.

# Indian Cities Population Dataset

## Dataset Description

This dataset contains demographic information about Indian cities, including population statistics and educational indicators.

## Business Importance

Population is one of the important factors influencing the future demand for EV charging stations. Cities with higher populations are more likely to require larger charging infrastructure.

## Expected Use in the Project

- Population Analysis
- Market Size Analysis
- Candidate City Evaluation
- Feature Engineering

# EV Charging Infrastructure Planning & Demand Prediction

## Notebook 01 : Population Data Preprocessing

### Business Objective

The objective of this notebook is to clean, preprocess, and standardize the population dataset for use in EV charging infrastructure planning. The cleaned dataset will serve as the foundation for EV growth prediction, charging demand estimation, waiting time prediction, and charging station recommendation models.

```python
# ==========================================================
#   Import Required Libraries
# ==========================================================

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 10)
```

```python
# ==========================================================
#   Load Population Dataset
# ==========================================================

population = pd.read_excel("../data/raw/population.xlsx")

print("Dataset Loaded Successfully")
```

```python
population["State"].nunique()
```

```python
population["City/District"].nunique()
```

```python
population.rename(columns={
    "City/District":"City"
}, inplace=True)
```

```python
population["State"] = population["State"].str.strip()

population["City"] = population["City"].str.strip()
```

```python
population["State"] = population["State"].str.title()

population["City"] = population["City"].str.title()
```

```python
population["Population"] = population["Population"].astype("int64")
```

```python
population.to_csv(
    "../data/processed/population_clean.csv",
    index=False
)

print("Population Dataset Saved Successfully")
```

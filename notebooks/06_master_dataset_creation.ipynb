## Notebook 06 : Master Dataset Creation

### Business Objective

The objective of this notebook is to merge all cleaned datasets into a single master dataset. This master dataset will be used for exploratory data analysis, feature engineering, machine learning model development, and dashboard creation.

```python
# ==========================================================
# Import Required Libraries
# ==========================================================

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

```python
# ==========================================================
# Load Cleaned Datasets
# ==========================================================

population = pd.read_csv("../data/processed/population_clean.csv")

ev = pd.read_csv("../data/processed/ev_registration_clean.csv")

station_master = pd.read_csv("../data/processed/station_master_clean.csv")

station_usage = pd.read_csv("../data/processed/station_usage_clean.csv")

charging = pd.read_csv("../data/processed/charging_stations_clean.csv")

print("All datasets loaded successfully.")
```

```python
print("Population :", population.shape)
print("EV Registration :", ev.shape)
print("Station Master :", station_master.shape)
print("Station Usage :", station_usage.shape)
print("Charging Stations :", charging.shape)
```

```python
station_master_city = (
    station_master
    .groupby(["State", "City"], as_index=False)
    .agg({
        "Total_Chargers": "sum",
        "AC_Chargers": "sum",
        "DC_Chargers": "sum",
        "Fast_Chargers": "sum",
        "Parking_Capacity": "sum",
        "Nearby_EV_Population": "mean"
    })
)
```

```python
station_usage_city = (
    station_usage
    .merge(
        station_master[["Station_ID", "State", "City"]],
        on="Station_ID",
        how="left"
    )
    .groupby(["Year", "Month", "State", "City"], as_index=False)
    .agg({
        "Avg_Daily_Users": "sum",
        "Peak_Hour_Users": "sum",
        "Avg_Wait_Time_Min": "mean",
        "Avg_Charging_Time_Min": "mean",
        "Queue_Length": "sum",
        "Energy_Delivered_kWh": "sum",
        "Sessions": "sum",
        "Completed_Sessions": "sum",
        "Cancelled_Sessions": "sum",
        "Revenue_INR": "sum",
        "Fault_Count": "sum",
        "Downtime_Hours": "sum",
        "Customer_Rating": "mean"
    })
)
```

```python
master = pd.merge(
    population,
    ev,
    on=["Year","State","City"],
    how="left"
)

print(master.shape)
```

```python
master = pd.merge(
    master,
    charging,
    on=["Year","State","City"],
    how="left"
)

print(master.shape)
```

```python
master = pd.merge(
    master,
    station_master,
    on=["State","City"],
    how="left"
)

print(master.shape)
```

```python
master = pd.merge(
    master,
    station_usage,
    on=["Station_ID", "Year", "Month"],
    how="left"
)
```
